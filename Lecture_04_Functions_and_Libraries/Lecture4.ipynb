## <span style="color:darkblue"> I. Built-in Functions </span>


<font size="4"> 

"Built-in" functions are those from Python libraries, e.g.

```print()```, ```type()```, ```round()```,```abs()```, ```len()```


- Inputs are known as "arguments"
- Outputs are known as "return values"


````python
### Function: type

# Argument: a Python object
# Return value: the type or class of the object

# Here, the float 3.14 is the argument
# We assign the return value to a variable we call q
q = type(3.14)
print(q)
````

````python
# we can also assign 3.14 to a variable, then use this as the argument:

my_num = 3.14
qqqq = type(my_num)
print(qqqq)
````

````python
# Here, we don't assign the return value to a variable, we just print it

my_string = "October 1"
print(type(my_string))
````

````python
#### Function: abs

# Argument: a single number (integer or floating point)
# Return value: a single number, which is the absolute value of the argument

tt = abs(-3)

y = 2.2
s = abs(y)

print(s)
print(tt)
print(abs(-10))
````

````python
# An "inappropriate" object for len would be a single number:

my_arg = 2
print(len(my_arg))
````

````python
#### Function: str

# Argument: Object
# Return value: Object converted to string

neg_num = -3
pos_num = abs(neg_num)
out_1 = str(pos_num)

out_2 = str(abs(neg_num) + 5)

print(out_1)
print(out_2)
````

````python
##### Function: round

# TWO arguments: one is optional.
# Argument 1: (required) a number
# Argument 2: (optional) number of digits to round to

# Return value: the rounded number. 
# (If the 2nd argument is omitted, rounded to an integer)

z = round(3.1415)
print(z)

zzz = round(3.1415, 3)
print(zzz)

````

````python
# it's important to think about imprecision in banker's rounding
%precision
````

````python
# The arguments of round are "named" (keywords)
help(round)
````

````python
# Valid ways of calling the round function
z1 = round(3.1415, 3)
z2 = round(3.1415, ndigits=3)
z3 = round(number=3.1415, ndigits=3)
z4 = round(ndigits=3, number=3.1415)

# invalid_output = round(number = 3.1415, 3) ## INVALID

print(z1)
print(z2)
print(z3)
print(z4)
````

````python
my_num = 3.1415
digits_I_want_to_keep = 3

z4 = round(my_num, digits_I_want_to_keep)
z5 = round(my_num, ndigits=digits_I_want_to_keep)
z6 = round(number=my_num, ndigits=digits_I_want_to_keep)

print(z4)
print(z5)
print(z6)
````

````python
help(print)
````

````python
##### Function: print

# Argument(s): 1 or more Python object. 
# (If more than one object, separate them by commas)

# Return values: None

t = print("Today", "is", "October", 1)
print(t) # This is None (print has no return value)
````

## <span style="color:darkblue"> II. List Methods</span>



### A method is like a function owned by a specific type of object
- #### List methods will allow us to add to, sort, and copy lists


##### Appending new elements to a list

````python
my_list = [7, 5, "Emory", -7, "Atlanta"]

# add new item to the end of the list using the "append" method.
my_list.append("Department of Data and Decision Sciences")
print(my_list)
````

````python
help(my_list.append)
````

##### Basic sorting of lists

````python
numerical_list = [5.3, 25, -0.1, -5, 11, 3, 1.2]
string_list = ["Frankie", "Nathan", "Peter", "Abed", "Britta", "Elroy"]

numerical_list.sort()
string_list.sort()

print(numerical_list)
print(string_list)
````

##### Who can explain the output of the next cell?

````python
a = ["left", "left"]
b = ["right", "right"]

temp = a.copy()
a[0] = b[0]
b[0] = temp[0]

print(a, b)
````

##### Copying lists

````python
# remember that the assignment operator doesn't make a copy if you put a list on the left-hand side
list_of_floats = [2.3, -1.0, 5.1, 9.9]

not_a_copy = list_of_floats

list_of_floats[0] = "Nothing"

print(not_a_copy)
````

````python
# if you want to make a copy, use list.copy() method

list_of_floats = [2.3, -1.0, 5.1, 9.9]

copy_of_list = list_of_floats.copy()

list_of_floats[0] = "Nothing"

print(copy_of_list)
````

##### Other list methods
See [here](https://docs.python.org/3/tutorial/datastructures.html) for all available list methods.
Here are two more useful ones

````python
# reverse the order of a list
colors = ["blue", "green", "red", "orange", "pink", "purple"]

colors.reverse()

print(colors)
````

````python
# count the number of times an item appears in the list
fav_colors = ["blue", "pink", "black", "pink", "red", "blue", "green", "black", "pink"]

pink_count = fav_colors.count("pink")

print(pink_count) # prints "3", the number of times "pink" appears in the list
````

## <span style="color:darkblue"> III. Importing Functions from Libraries </span>


### The `math` library


##### The math library holds mathematical functions that are used often enough to be worth writing once, but not often enough to be built into Python itself.

````python
import math


# The math library has a square root function
print(math.sqrt(25))

# It also has factorial
print(5 * 4 * 3 * 2 * 1)
print(math.factorial(5))
````

````python

# A library can also store useful VALUES, not just functions.
# math.pi is not a function -- it is just a float that lives inside the library:
print(math.pi)
print(type(math.pi))

# We can hand it straight to the built-in round() from Section I:
print(round(math.pi, 3))
````

#### You can see all of the `math` library functions and constants in the [documentation](https://docs.python.org/3/library/math.html)

### The `statistics` library

The statistics library summarizes a lists of numbers.

````python
print(numerical_list)
````

````python
import statistics

# numerical_list is the list we sorted back in Section II
print(numerical_list)

print(statistics.mean(numerical_list))
print(statistics.median(numerical_list))
print(statistics.stdev(numerical_list))
````

#### You can look at the [documentation](https://docs.python.org/3/library/statistics.html) for all of the statistics functions available.

### The `numpy` library
<font size="4"> 

Numpy (short for "numerical Python") allows us to perform mathematical operations on multiple values at once. Nearly every other scientific library in Python is built on top of and for compatability with numpy. An example of a library built on top of numpy is scipy, which we will see in the next lecture.


**Numpy is designed to efficiently perform calculations on multiple values.** This is typically done with a special type of object called an array, but for now we will demonstrate with lists.

The numpy library is vast. Here is the numpy [documentation](https://numpy.org/doc/stable/reference/routines.html)

````python
# the math library gives an error if we try to apply the square root function to multiple values

side_areas = [4, 9, 16, 25]

print(math.sqrt(side_areas)) 
````

````python
import numpy as np

# numpy's version of sqrt works just fine on a list
print(np.sqrt(side_areas))
````

note: we often import numpy **as** `np` to avoid writing out `numpy` in its entirety every time we use a function from the library.

````python
# here are a few more numpy functions  applied to multiple values

# sum
print(np.sum([1, 2, 3, 4]))
# product
print(np.prod([1, 2, 3, 4]))
# cumulative sum
print(np.cumsum([1, 2, 3, 4]))
# cumulative product
print(np.cumprod([1, 2, 3, 4]))
# mean
print(np.mean([1, 2, 3, 4]))
````

````python
# finding maximum and minimum values

quiz_scores = np.array([7.5, 9.0, 4.5, 8.0, 6.5])

print(np.max(quiz_scores))
print(np.min(quiz_scores))
````

`numpy` contains almost all of the functions and constants from `math` and `statistics` along with many more operations. However there are exceptions:

````python
print(np.prod([5, 4, 3, 2, 1]))
print(math.factorial(5))

print(np.factorial(5))
````

We will learn more about these libraries in future lectures.

For now, here is a quick illustration of some specialized functions that they contain:


````python
# specialized function from the pandas library for loading a .csv file into our environment
# The csv we are loading has been pre-filtered to only contain five openings
df = pd.read_csv('data/JuditPolgar_white_top5.csv')

# put all the rows that share the same opening into the same group
groups = df.groupby('black_move_1')

# from each group, look at only the result column
results = groups['result']

# average the results in each group -> that opening's win rate
win_rates = results.mean()

# turn the grouped answer back into a normal table with two columns: opening, result
win_rates = win_rates.reset_index()

# one bar per opening: names along the bottom, win rate as the height
plt.barh(win_rates['black_move_1'], win_rates['result'])
plt.show()
````
